### Section 1. 필터링, 최소값, 중앙값

- 'f5' 컬럼이 0이 아닌 데이터(행)를 구하시오.
- 앞에서 구한 데이터에 ‘views’ 컬럼 결측치를 ‘views’ 컬럼의 최솟값으로 채워주세요.
- 그리고 ‘views’ 컬럼의 중앙값을 계산해 정수로 구하시오.

````python
import pandas as pd
# df = pd.read_csv("type1_data1.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/type1_data1.csv")

# 강의를 보기 전에 먼저 풀어보세요!
# 혼자 풀었음.

# 0 제외
df = df[df['f5'] != 0]

# 결측치 채우기
df['views'] = df['views'].fillna(df['views'].min())

# 중앙값 계산
print(int(df['views'].median()))
````

````python
import pandas as pd
# df = pd.read_csv("type1_data1.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/type1_data1.csv")

# 0 제외
# print(df.shape)
cond = df['f5'] != 0
df = df[cond]
# print(df.shape)

# 'views' 컬럼 결측치 최소값으로 채우기
# print(df['views'].isnull().sum())
mi = df['views'].min()
df['views'] = df['views'].fillna(mi)
# print(df['views'].isnull().sum())

# 'views' 컬럼 중앙값 계산
print(int(df['views'].median()))
````

````python
map()을 활용한 풀이
pythonimport pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/inf/refs/heads/main/p1/type1_data1.csv")

# 1 중복 데이터 제거
df = df.drop_duplicates()

# 2 f3 변환 (결측치 -> 0, silver -> 1, gold -> 2, vip -> 3)
df['f3'] = df['f3'].map({'silver': 1, 'gold': 2, 'vip': 3}).fillna(0)

# 3 f3 sum()
result = df['f3'].sum()
print(result)

map() 핵심 개념
map()은 딕셔너리를 넘겨주면 해당 값으로 치환하고, 딕셔너리에 없는 값은 자동으로 NaN으로 처리합니다. 따라서 결측치(NaN)와 매핑되지 않은 값 모두 .fillna(0)으로 한 번에 0으로 채울 수 있어 코드가 훨씬 간결해집니다.
기존 코드처럼 조건문(cond1, cond2, cond3)으로 작성해도 답은 동일하게 나오지만, map()을 사용하면 한 줄로 처리할 수 있어 시험 시간을 절약할 수 있습니다.
````

````python
import pandas as pd
# df = pd.read_csv("type1_data1.csv")
df = pd.read_csv("https://raw.githubusercontent.com/lovedlim/bigdata_analyst_cert/main/part1/ch3/type1_data1.csv")

# 자료형이 object 컬럼 선택
cols = df.select_dtypes(exclude='object').columns
df = df[cols]

# 결측치 0으로 대체
df = df.fillna(0)

# 행과 열 변경
df = df.T

# 행별 합이 3000이상인 데이터 수
print(sum(df.sum() > 3000))
````
