## Section1. 데이터프레임과 시리즈

```python
# 판다스 라이브러리 불러오기
import pandas as pd
```

```python
# 시리즈 만들기 (문자열)
menu = pd.Series(['비빔밥', '김치찌개', '된장찌개'])
menu
```

```python
# 시리즈 만들기(정수형)
price = pd.Series([10000, 9000, 8000])
price
```

```python
# 데이터 프레임 만들기 pd.DataFrame({"컬럼명":데이터})
pd.DataFrame({
    '메뉴': menu,
    '가격': price
})
```

```python
# 데이터 프레임 만들기 pd.DataFrame({"컬럼명":데이터})
df = pd.DataFrame({
    "메뉴":['비빔밥', '김치찌개', '된장찌개'],
    "가격":[10000, 9000, 8000],
    "원산지":['국내산', '국내산', '국내산']
})
df
```

```python
# 시리즈 선택 (메뉴)
df['메뉴']
```

```python
# 데이터프레임 선택 (메뉴)
df[['메뉴']]
```

```python
# 데이터프레임 선택 (메뉴, 가격)
df[['메뉴', '가격']]
```

```python
# 데이터프레임 선택 (메뉴, 가격)
cols = ['메뉴', '가격']
df[cols]
```

```python
# 데이터 프레임 type 확인
print("df          : ", type(df))
print("df['가격']   : ", type(df['가격']))
print("df[['가격']] : ", type(df[['가격']]))
```

## Section2. 데이터 저장 및 불러오기

```python
# 데이터 준비
df = pd.DataFrame({
    "메뉴":['아메리카노', '카페라떼', '카페모카', '카푸치노', '에스프레소', '밀크티', '녹차'],
    "가격":[4500, 5000, 5500, 5000, 4000, 5900, 5300],
    "칼로리":[10, 110, 250, 110, 20, 210, 0],
})
df
```

```python
# 저장 temp.csv
df.to_csv('temp.csv')
```

```python
# 저장한 데이터 불러오기
temp_df = pd.read_csv('temp.csv')
temp_df.head()
```

```python
# 데이터 불러오기
df.to_csv('temp.csv', index=False)
temp_df = pd.read_csv('temp.csv')
temp_df.head()
```

## Section3. 탐색적 데이터 분석(EDA)

```python
# 데이터 샘플 확인: 앞에서 부터 n개 (기본 5개)
df.head(2)
```

```python
# 데이터 샘플 확인: 뒤에서 부터 n개 (기본 5개)
df.tail(3)
```

```python
# sample(랜덤)
df.sample(3)
```

```python
# 데이터 프레임 크기 (행, 컬럼)
df.shape
```

```python
# 컬럼 형태(type)
df.info()
```

```python
# 상관관계
df.corr(numeric_only=True)
```

```python
# 중복 값이 있는 데이터 생성
df_car = pd.DataFrame({
    "car":['Sedan','SUV','Sedan','SUV','SUV','SUV','Sedan','Sedan','Sedan','Sedan','Sedan'],
    "size":['S','M','S','S','M','M','L','S','S', 'M','S']
})
df_car.head(3)
```

```python
# 항목 종류 수
df_car.nunique()
```

```python
# 항목 종류
print(df_car['car'].unique())
print(df_car['size'].unique())
```

```python
# 항목별 개수
print(df_car['car'].value_counts())
print(df_car['size'].value_counts())
```

```python
# 기초 통계 (숫자)
df.describe()
```

```python
# 기초 통계 (문자)
df_car.describe(include="O")    # 영어 대문자 O. 숫자 0 아님.
```

```python
# 기초 통계 (문자)
df.describe(include="O")
```
