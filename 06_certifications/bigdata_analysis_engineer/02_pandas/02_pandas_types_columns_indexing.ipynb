```python
import pandas as pd
```

## Section4. 자료형 변환

```python
# 데이터 준비
data = {
    "메뉴":['아메리카노', '카페라떼', '카페모카', '카푸치노', '에스프레소', '밀크티', '녹차'],
    "가격":[4500.0, 5000.0, 5500.0, 5000.0, 4000.0, 5900.0, 5300.0],
    "칼로리":['10', '110', '250', '110', '20', '210', '0'],
}
df = pd.DataFrame(data)
df.info()
```

```python
# 자료형 변환 / astype /  float -> int
df['가격'] = df['가격'].astype(int)
df.info()
```

```python
# 자료형 변환 / astype /  object -> float
df['칼로리'] = df['칼로리'].astype(float)
df.info()
```

## Section5. 새로운 컬럼 추가

```python
# 데이터 준비
df = pd.DataFrame({
    "메뉴":['아메리카노', '카페라떼', '카페모카', '카푸치노', '에스프레소', '밀크티', '녹차'],
    "가격":[4500, 5000, 5500, 5000, 4000, 5900, 5300],
    "칼로리":[10, 110, 250, 110, 20, 210, 0],
})
df.to_csv('cafe.csv', index=False)
```

```python
# 데이터 불러오기
df = pd.read_csv('cafe.csv')
df.head()
```

```python
# 새로운 컬럼 추가 (값은 0)
df['new'] = 0
df.head()
```

```python
# 할인가 컬럼 추가
discount = 0.2
df['할인가'] = df['가격'] * (1-discount)
df.head()
```

## Section6. 데이터 삭제 (컬럼, 행)

```python
# 데이터 불러오기
df = pd.read_csv('cafe.csv')
df.head(3)
```

### 행 삭제

```python
# 행 삭제
df = df.drop(1, axis=0)
df.head(3)
```

### 컬럼 삭제

```python
# 데이터 삭제
# axis=0:행방향, axis=1:열방향(컬럼)
df = df.drop("칼로리", axis=1)
df.head(3)
```

## Section7. 인덱싱/슬라이싱(loc)
- 인덱스 명
- 인덱스 명(범위), 컬럼 명(범위)

```python
# 데이터 불러오기
df = pd.read_csv('cafe.csv')
df.head(2)
```

```python
# 인덱싱 (행 전체)
# 아메리카노
df.loc[0]
```

```python
# 카페라떼 가격
df.loc[1, '가격']
```

```python
# 슬라이싱 (컬럼 전체)
# 가격
df.loc[:, '가격']
```

```python
# 슬라이싱
df.loc[:, '메뉴':'칼로리']
```

```python
# 슬라이싱
# 카페모카 메뉴와 칼로리
df.loc[2, ['메뉴', '칼로리']]
```

```python
# 슬라이싱
df.loc[1:3, '메뉴':'가격']
```

## Section8. 인덱싱/슬라이싱(iloc)
- 인덱스 번호
- 인덱스 번호(범위), 컬럼 번호(범위)

```python
# 데이터 불러오기
df = pd.read_csv('cafe.csv')
df = df.drop(0, axis=0) # 첫번째 행 삭제
df.head(2)
```

```python
# 인덱싱 (행 전체)
# 카페라떼
df.iloc[0]
```

```python
# 슬라이싱 (컬럼 전체)
# 가격
df.iloc[:, 1]
```

```python
# 슬라이싱
# 카푸치노 메뉴와 가격
df.iloc[2, :2]
```

```python
# 슬라이싱 (카페모카와 카푸치노의 전체 데이터)
df.iloc[1:3, :]
```

```python
# loc 슬라이싱 (카페모카와 카푸치노의 전체 데이터)
df.loc[2:3, '메뉴':'칼로리']
```
