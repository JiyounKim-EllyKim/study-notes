## Section9. 데이터 추가/변경

```python
# 데이터 준비
df = pd.DataFrame({
    "메뉴":['아메리카노', '카페라떼', '카페모카', '카푸치노', '에스프레소', '밀크티', '녹차'],
    "가격":[4500, 5000, 5500, 5000, 4000, 5900, 5300],
    "칼로리":[10, 110, 250, 110, 20, 210, 0],
})
df.to_csv('cafe.csv', index=False)
```

```python
# 데이터 불러오기
df = pd.read_csv('cafe.csv')
df.head(2)
```

```python
# 결측값으로 추가, 원두 컬럼을 만들고 결측값(NaN)으로 대입
import numpy as np
df['원산지'] = np.nan
df.head(4)
```

```python
# 아메리카노의 원두는 '콜롬비아' 데이터 추가
df.loc[0, '원산지'] = '콜롬비아'
df.loc[2:3, '원산지'] = '과테말라'
df.head()
```

```python
# 리스트 형태로 데이터(행) 추가
df.loc['시즌'] = ['크리스마스라떼', 6000, 300, '한국']
df.tail()
```

```python
# 딕셔너리 형태로 데이터(행) 추가
df.loc[7] = {'메뉴':'딴짓커피', '가격':2000, '칼로리':20}
df.tail()
```

```python
# cafe2 저장
df = df.drop("시즌", axis=0)
df.to_csv('cafe2.csv', index=False)
```

## Section10. 정렬

```python
# 데이터 불러오기
df = pd.read_csv('cafe2.csv')
df.head(2)
```

```python
# 인덱스 기준 (기본값 ascending=True)
df.sort_index(ascending=False)
```

```python
# 값 기준 (기본값 ascending=True)
# 메뉴 기준 정렬
df.sort_values('가격', ascending=False)
```

```python
# 가격과 메뉴 기준 정렬
df = df.sort_values(['가격', '메뉴'], ascending=[False, True])
df
```

```python
# 인덱스 새로 만들기 drop=True
df = df.reset_index(drop=True)
df
```

## Section11. 필터링

```python
# 데이터 불러오기
df = pd.read_csv('cafe2.csv')
df.head(2)
```

```python
df['칼로리'] < 50
```

```python
# 1개 조건
cond = df['칼로리'] < 50
df[cond]
```

```python
# 1개 조건
df[df['칼로리'] < 50]
```

```python
# 칼로리 50 이상
cond = df['칼로리'] < 50
df[~cond]
```

```python
# 2개 이상 일때 (AND)
cond1 = df['가격'] >= 5000
cond2 = df['칼로리'] > 100
df[cond1 & cond2]
```

```python
# 2개 이상 일때 (OR)
df[cond1 | cond2]
```

```python
# 문자열 1개 조건
# 원두 == 콜롬비아
cond = df['원산지'] == '콜롬비아'
df[cond]
```

```python
# 문자열과 숫자 조건 (AND)
cond1 = df['원산지'] == '과테말라'
cond2 = df['가격'] <= 5000
df[cond1 & cond2]
```

```python
# isin
df['메뉴'].isin(['녹차'])
```

```python
# isin 활용
cond = df['메뉴'].isin(['녹차'])
df[cond]
```

```python
# isin 활용 (복수 컬럼)
box = ['녹차', '카푸치노', '카페라떼']
cond = df['메뉴'].isin(box)
df[cond]
```

## Section12. 결측치 처리

```python
# 데이터 불러오기
df = pd.read_csv('cafe2.csv')
df.head(2)
```

```python
# 결측치 여부
df.isnull()
```

```python
# 컬럼별 결측치 확인
df.isnull().sum()
```

```python
# 컬럼별 결측치 확인
df.isna().sum()
```

```python
# 결측값 채우기
df['원산지'] = df['원산지'].fillna('코스타리카')
df
```

```python
# 저장
df.to_csv("cafe3.csv", index=False)
```

## Section13. 값 변경

```python
# 데이터 확인
pd.read_csv("cafe3.csv")
df.head(3)
```

```python
# 문자 변경 : 아메리카노 -> 룽고, 녹차 -> 그린티
df = df.replace('아메리카노', '룽고')
df = df.replace('녹차', '그린티')
df
```

```python
# 딕셔너리 활용
change = {'룽고':'아메리카노', '그린티':'녹차'}
df = df.replace(change)
df
```

```python
# loc로 값 변경하기
df.loc[6, '원산지'] = '대한민국'
df.tail(3)
```

```python
# 이벤트가 전체 1000으로 변경
df.loc[1:2, '이벤트가'] = 1000
df.head()
```

```python
# map() 함수
import pandas as pd
df = pd.DataFrame({'만족도': ['만족', '보통' ,'불만', '보통', '만족']})

mapping = {'만족': 3, '보통': 2, '불만': 1}
df['만족도_숫자'] = df['만족도'].map(mapping)
df
```
