## Section14. 문자열

```python
import pandas as pd
df = pd.DataFrame({'A': ['데이터 분석', '기본 학습서', '퇴근 후 열공'],
                   'B': [10, 20, 30],
                   'C': ['ab cd', 'AB CD', 'ab cd ']
                   })
df
```

```python
# 값 변경 시도
df['A'] = df['A'].replace('분석','시각화')
df
```

```python
# 값 변경(str)
df['A'] = df['A'].str.replace('분석','시각화')
df
```

```python
# 값 변경 (숫자)
df['B'] = df['B'].replace(10, 100)
df
```

```python
# 에러
# df['B'] = df['B'].str.replace(20, 200)
# print(df)
```

```python
# 문자열 분리
df['A'].str.split()
```

```python
# 첫번째 행 데이터
df['A'].str.split()[0]
```

```python
# 각 행별 첫번째 요소 추출
df['D'] = df['A'].str.split().str[0]
df
```

```python
# 문자 검색
df['A'].str.contains('분석')
```

```python
# 문자 포함 여부
df['기본포함유무'] = df['A'].str.contains('기본')
df
```

```python
# isin: 완전히 동일해야 됨. 여러 단어 필터링 가능.
menu = pd.Series(['맛난버거 세트', '매운 치킨버거', '더블 치즈버거'])
menu.isin(['맛난버거 세트', '더블 치즈버거'])
```

```python
# contains: 일부만 포함되어도 됨. 다만 한 단어만 필터링 가능.
menu.str.contains('세트')
```

```python
# 문자 길이
df['문자길이'] = df['A'].str.len()
df
```

```python
# 대소문자 구분
'AB cd' == 'ab cd'
```

```python
# 소문자로 변경
df['C'] = df['C'].str.lower()
df['C']
```

```python
# 대분자로 변경
df['C'] = df['C'].str.upper()
df['C']
```

```python
# 공백
df['C'] == 'AB CD'
```

```python
# 공백 제거 및 소문자 처리
df['C'] = df['C'].str.lower()
df['C'] = df['C'].str.replace(' ','')
df['C']
```

```python
# 문자열 슬라이싱
df['C'].str[1:3]
```

```python
# 행 선택
df['C'][1:3]
```

## Section15. 내장함수

```python
# 데이터 불러오기
df = pd.read_csv("cafe4.csv")
df.head(2)
```

```python
# 행의 수
print(len(df))
print(df.shape[0])
```

```python
# 조건
cond = df['가격'] > 5000

# sum으로 조건에 맞는 개수 구하기
print(sum(cond))

# len으로 조건에 맞는 개수 구하기
print(len(df[cond]))
```

```python
# 합계 (컬럼)
df.sum(numeric_only=True)
```

```python
# 합계 (행)
df.sum(axis=1, numeric_only=True)
```

```python
# 최대, 최소, 평균, 중앙, 합계, 표준편차, 분산
print("최대값: ", df['가격'].max())
print("최소값: ", df['가격'].min())
print("평균값: ", df['가격'].mean())
print("중앙값: ", df['가격'].median())
print("합계: ", df['가격'].sum())
print("표준편차: ", df['가격'].std())
print("분산: ", df['가격'].var())
```

```python
# 분위수
print("1분위수", df['가격'].quantile(0.25))
print("2분위수", df['가격'].quantile(0.5))
print("3분위수", df['가격'].quantile(0.75))
```

```python
# 하위 25% 데이터
cond = df['가격'] <= df['가격'].quantile(0.25)
df[cond]
```

```python
# 상위 25% 데이터
cond = df['가격'] >= df['가격'].quantile(0.75)
df[cond]
```

```python
# 최빈값 구하기
df['원산지'].mode()[0]
```

```python
# 최대값의 인덱스
df['가격'].idxmax()
```

```python
# 최대값의 인덱스로 데이터 선택
max_idx = df['가격'].idxmax()
df.loc[max_idx]
```

```python
# 최대값의 인덱스 활용
df.loc[max_idx, '메뉴']     # 추천 방법
```

```python
# 가격 컬럼 중 가장 큰 값 3개
df.nlargest(3, '가격')
```

```python
# 가격 컬럼 중 가장 작은 값 2개
df.nsmallest(2, '가격')
```

```python
# apply 적용해서 새로운 컬럼 생성 (칼로리 컬럼 활용)
def cal(x):
    if x >= 100:
        return "No"
    else:
        return "Yes"

df['먹어도 될까요'] = df['칼로리'].apply(cal)
df
```

```python
import pandas as pd
df = pd.DataFrame({'Name': {0: '쿼카', 1: '알파카', 2: '시바견'},
                   '수학': {0: 90, 1: 93, 2: 85},
                   '영어': {0: 92, 1: 84, 2: 86},
                   '국어': {0: 91, 1: 94, 2: 83},})

df
```

```python
# melt
pd.melt(df, id_vars=['Name'])
```

```python
# melt - value 선택
pd.melt(df, id_vars=['Name'], value_vars=['수학', '영어'])
```

```python
# melt 컬럼명 설정
pd.melt(df, id_vars=['Name'], var_name='과목', value_name='점수')
```
