## 반복문

```python
# 0부터 4까지 반복 출력

for i in range(5):
    print(i)
```

```python
# 5부터 9까지 반복 출력

for i in range(5, 10):
    print(i)
```

```python
# 리스트에 있는 값, 반복문을 통해 출력
listbox = [2,4,6,8,10]

for i in listbox:
    print(i)
```

```python
# 리스트에 있는 값 반복문을 통해 +1 한 값, 출력
listbox = [2,4,6,8,10]

for i in listbox:
    print(i+1)
```

```python
# 리스트에 있는 문자열 출력
listbox = ['네모','세모','동그라미']

for i in listbox:
    print(i)
```

```python
# 리스트에 있는 문자열과 인덱스값 출력
listbox = ['네모','세모','동그라미', '별']

for idx, item in enumerate(listbox):
    print(idx, item)
```

```python
# Tip (print(f스트링))
for idx, item in enumerate(listbox):
    print(f'인덱스 {idx}인 값은 {item}입니다.')
```

### 함수(리턴 값)

```python
# 함수 정의 (리턴 값)
def plus(x, y):
    result = x + y
    return result
```

```python
# 함수 호출
plus(2, 3)
```

```python
# 리턴 값 받기
result = plus(2, 3)
print(result)
```

```python
# 리턴값이 2개 (최소, 최대값을 구하는 함수)
listbox = [15, 46, 78, 24, 56]

def min_max(data):
    min_value = min(data)
    max_value = max(data)
    return min_value, max_value

min_value, max_value = min_max(listbox)
print(min_value, max_value)
```

```python
# 평균을 구하는 함수
listbox = [15, 46, 78, 24, 56]

def average(data):
    avg = sum(data) / len(data)
    return avg

avg = average(listbox)
print(avg)
```

### Quiz

```python
# 주어진 리스트의 최대값 - 최소값을 구하는 함수를 만드시오
# listbox = [15, 46, 78, 24, 56]

def diff_max_min(data):
    min_value = min(data)
    max_value = max(data)
    return max_value - min_value

listbox = [15, 46, 78, 24, 56]
result = diff_max_min(listbox)
print(result)
```
