## 문자열

### 문자열 변경

```python
# 한 단어 변경
text = '빅데이터 분석 기사 파이썬 공부'
text = text.replace('공부', '스터디')
text
```

```python
# 여러 단어 변경
text = '빅데이터 분석 기사 파이썬 공부'
text = text.replace('파이썬', '머신러닝').replace('공부', '스터디')
text
```
