## 조건문

- 들여쓰기: tab
- 내어쓰기: shift + tab

```python
# True 조건
if True:
    print("참입니다.")
```

```python
# False 조건
if False:
    print("실행 안 됨")
```

```python
box = 10

if box < 5:
    print("box는 5보다 작다.")
```

```python
box = 10

if box == 10:
    print('box는 10입니다.')
    print(box)
else:
    print('box는 10이 아닙니다.')
```

```python
box = 15

if box >= 10:
    print("box는 10 이상입니다.")
else:
    print("box는 10 미만입니다.")
```

```python
box = 7

if box >= 10:
    print("box는 10 이상입니다.")
elif box < 10 and box >= 5:
    print("box는 5 이상 10 미만입니다.")
else:
    print("box는 5 미만입니다.")
```
